```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('parrot.jpg', cv2.IMREAD_GRAYSCALE)
```

```python
plt.figure(figsize=(14,10))   
plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')
```

```python
plt.figure(figsize=(14,10))   # Increased size
plt.subplot(2,2,2)
plt.hist(img.ravel(), bins=256, range=[0,256])
plt.title('Histogram of Original Image')
```

```python
equalized_img = cv2.equalizeHist(img)
```

```python

plt.figure(figsize=(14,10))   # Increased size
plt.subplot(2,2,3)
plt.imshow(equalized_img, cmap='gray')
plt.title('Equalized Image')
plt.axis('off')
```

```python
plt.figure(figsize=(14,10))   # Increased size
plt.subplot(2,2,4)
plt.hist(equalized_img.ravel(), bins=256, range=[0,256])
plt.title('Histogram of Equalized Image')
```

```python
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('parrot.jpg')
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(14,10)) 
plt.subplot(2,2,1)
plt.imshow(img_rgb)
plt.title('Original Color Image')
plt.axis('off')
```

```python
plt.figure(figsize=(14,10)) 
plt.subplot(2,2,2)
color = ('b','g','r')
for i, col in enumerate(color):
    hist = cv2.calcHist([img], [i], None, [256], [0,256])
    plt.plot(hist, color=col)
plt.title('Histogram of Original Image')
```

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
```

```python
hsv[:,:,2] = cv2.equalizeHist(hsv[:,:,2])
```

```python
enhanced_img = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)
```

```python
enhanced_rgb = cv2.cvtColor(enhanced_img, cv2.COLOR_BGR2RGB)
```

```python
plt.figure(figsize=(14,10)) 
plt.subplot(2,2,3)
plt.imshow(enhanced_rgb)
plt.title('Enhanced Color Image')
plt.axis('off')
```

```python
plt.figure(figsize=(14,10)) 
plt.subplot(2,2,4)
for i, col in enumerate(color):
    hist = cv2.calcHist([enhanced_img], [i], None, [256], [0,256])
    plt.plot(hist, color=col)
plt.title('Histogram of Enhanced Image')
```
